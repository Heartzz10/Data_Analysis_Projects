```python
import pandas as pd
table = pd.read_excel('Final.xlsx')
table
```

```python
copy_table = table.copy()
data = table.drop(['OBJECTID', 'Name', 'SUM_Pop_Estimate', 'Shape_Length', 'Shape_Area'], axis = 1)
data
```

```python
colomn = data.columns.tolist()
colomn
```

```python
data_normalized = data[colomn].apply(lambda x: (x - x.min()) / (x.max() - x.min()))
data_normalized
```

```python
weights = {
    'Pop_den': 0.02941,            
    'Medical_percent': 0.14706,
    'Post_percent': 0.02941,
    'Service_percent': 0.08824,
    'Police_percent': 0.14706,
    'Market_Count': 0.08824,
    'Food_Count':0.14705,
    'Shop_Count': 0.14706,
    'Leisure_Count': 0.08824,
    'Trip_Count': 0.02941,
    'Rent_price': -0.02941,
    'Noise': -0.02941
}
sum(abs(value) for value in weights.values())
```

```python
data_normalized['Sum_Score'] = sum(data_normalized[col] * weight for col, weight in weights.items())
data_normalized['Sum_Score']
```

```python
result = pd.DataFrame({
    'OBJECTID': table['OBJECTID'],
    'Name':table['Name'],
    'Score': data_normalized['Sum_Score']
})
result
```

```python
result.to_excel('Result_Score.xlsx', index = False)
```
